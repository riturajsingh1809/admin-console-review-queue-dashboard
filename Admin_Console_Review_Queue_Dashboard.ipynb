```python
# =====================================================
# TASK 18 : Admin Console & Review Queue
# PART 1 : Import Libraries & Load Dataset
# =====================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display all columns
pd.set_option("display.max_columns", None)

# Upload Excel File
from google.colab import files
uploaded = files.upload()

# Read Dataset
df = pd.read_excel("Sample Dataset.xlsx")

# First Five Records
print("First Five Records")
display(df.head())

# Last Five Records
print("Last Five Records")
display(df.tail())

# Shape
print("Dataset Shape")
print(df.shape)

# Column Names
print("Columns")
print(df.columns)

# Information
print("Dataset Information")
df.info()

# Data Types
print(df.dtypes)

# Missing Values
print("Missing Values")
print(df.isnull().sum())

# Duplicate Records
print("Duplicate Records")
print(df.duplicated().sum())

# Statistical Summary
display(df.describe(include='all'))

print("Part 1 Completed Successfully")
```

```python
# =====================================================
# TASK 18 : Admin Console & Review Queue
# PART 2 : Data Cleaning & Analysis
# =====================================================

import pandas as pd

# ----------------------------
# Missing Values
# ----------------------------
print("Missing Values")
display(df.isnull().sum())

# Fill Missing Values (if any)
df.fillna("Not Available", inplace=True)

# ----------------------------
# Remove Duplicate Records
# ----------------------------
duplicates = df.duplicated().sum()
print("Duplicate Records :", duplicates)

df.drop_duplicates(inplace=True)

# ----------------------------
# Dataset Shape After Cleaning
# ----------------------------
print("Dataset Shape After Cleaning")
print(df.shape)

# =====================================================
# Status Analysis
# =====================================================

print("\nCandidate Status Analysis")
status_count = df["Status"].value_counts()

display(status_count)

# =====================================================
# Review Queue Analysis
# =====================================================

print("\nReview Status Analysis")

review_count = df["Review_Status"].value_counts()

display(review_count)

# =====================================================
# Recruiter-wise Analysis
# =====================================================

print("\nRecruiter-wise Candidates")

recruiter_analysis = df.groupby("Recruiter_Name").size().sort_values(ascending=False)

display(recruiter_analysis)

# =====================================================
# Company-wise Analysis
# =====================================================

print("\nCompany-wise Hiring")

company_analysis = df.groupby("Company").size().sort_values(ascending=False)

display(company_analysis)

# =====================================================
# Job Title Analysis
# =====================================================

print("\nJob Title Analysis")

job_analysis = df["Job_Title"].value_counts()

display(job_analysis)

# =====================================================
# Location Analysis
# =====================================================

print("\nLocation Analysis")

location_analysis = df["Location"].value_counts()

display(location_analysis)

# =====================================================
# Priority Analysis
# =====================================================

print("\nPriority Analysis")

priority_analysis = df["Priority"].value_counts()

display(priority_analysis)

# =====================================================
# Average Recruiter Rating
# =====================================================

print("\nAverage Recruiter Rating")

rating_analysis = df.groupby("Recruiter_Name")["Rating"].mean().sort_values(ascending=False)

display(rating_analysis)

# =====================================================
# Experience Analysis
# =====================================================

print("\nExperience Statistics")

display(df["Experience"].describe())

# =====================================================
# Top Recruiters
# =====================================================

print("\nTop Recruiters")

top_recruiters = (
    df.groupby("Recruiter_Name")
      .size()
      .reset_index(name="Total_Candidates")
      .sort_values(by="Total_Candidates", ascending=False)
)

display(top_recruiters)

print("✅ Part 2 Completed Successfully")
```

```python
# =====================================================
# TASK 18 : Admin Console & Review Queue
# PART 3 : Data Visualization
# =====================================================

import matplotlib.pyplot as plt

# -----------------------------
# Chart 1 : Candidate Status
# -----------------------------
status_count = df["Status"].value_counts()

plt.figure(figsize=(7,5))
status_count.plot(kind="bar")
plt.title("Candidate Status")
plt.xlabel("Status")
plt.ylabel("Number of Candidates")
plt.xticks(rotation=0)
plt.show()


# -----------------------------
# Chart 2 : Review Queue
# -----------------------------
review_count = df["Review_Status"].value_counts()

plt.figure(figsize=(6,6))
review_count.plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Review Queue Distribution")
plt.ylabel("")
plt.show()


# -----------------------------
# Chart 3 : Company-wise Hiring
# -----------------------------
company_analysis = df["Company"].value_counts()

plt.figure(figsize=(8,5))
company_analysis.plot(kind="bar")
plt.title("Company-wise Hiring")
plt.xlabel("Company")
plt.ylabel("Candidates")
plt.xticks(rotation=45)
plt.show()


# -----------------------------
# Chart 4 : Top Recruiters
# -----------------------------
top_recruiters = df["Recruiter_Name"].value_counts()

plt.figure(figsize=(8,5))
top_recruiters.plot(kind="bar")
plt.title("Top Recruiters")
plt.xlabel("Recruiter")
plt.ylabel("Candidates Reviewed")
plt.xticks(rotation=45)
plt.show()


# -----------------------------
# Chart 5 : Priority Analysis
# -----------------------------
priority = df["Priority"].value_counts()

plt.figure(figsize=(6,4))
priority.plot(kind="bar")
plt.title("Priority Distribution")
plt.xlabel("Priority")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()


# -----------------------------
# Chart 6 : Job Title Analysis
# -----------------------------
jobs = df["Job_Title"].value_counts()

plt.figure(figsize=(8,5))
jobs.plot(kind="bar")
plt.title("Job Title Distribution")
plt.xlabel("Job Title")
plt.ylabel("Open Positions")
plt.xticks(rotation=30)
plt.show()


# -----------------------------
# Chart 7 : Recruiter Rating
# -----------------------------
rating = df.groupby("Recruiter_Name")["Rating"].mean()

plt.figure(figsize=(8,5))
rating.plot(kind="bar")
plt.title("Average Recruiter Rating")
plt.xlabel("Recruiter")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()


# -----------------------------
# Chart 8 : Experience Distribution
# -----------------------------
plt.figure(figsize=(7,5))
plt.hist(df["Experience"], bins=10)
plt.title("Candidate Experience Distribution")
plt.xlabel("Experience (Years)")
plt.ylabel("Number of Candidates")
plt.show()


print("✅ Part 3 Completed Successfully")
```

```python
# =====================================================
# TASK 18 : Admin Console & Review Queue
# PART 4 : KPI Dashboard & Export Report
# =====================================================

import pandas as pd

# =====================================================
# KPI SUMMARY
# =====================================================

total_candidates = len(df)
total_recruiters = df["Recruiter_Name"].nunique()
total_companies = df["Company"].nunique()
total_jobs = df["Job_Title"].nunique()

pending_reviews = (df["Review_Status"] == "Pending").sum()
completed_reviews = (df["Review_Status"] == "Completed").sum()

selected_candidates = (df["Status"] == "Selected").sum()
rejected_candidates = (df["Status"] == "Rejected").sum()
shortlisted_candidates = (df["Status"] == "Shortlisted").sum()
in_review_candidates = (df["Status"] == "In Review").sum()

average_rating = round(df["Rating"].mean(), 2)
average_experience = round(df["Experience"].mean(), 2)

print("="*55)
print("        ADMIN CONSOLE DASHBOARD SUMMARY")
print("="*55)

print(f"Total Candidates      : {total_candidates}")
print(f"Total Recruiters      : {total_recruiters}")
print(f"Total Companies       : {total_companies}")
print(f"Total Job Roles       : {total_jobs}")
print(f"Pending Reviews       : {pending_reviews}")
print(f"Completed Reviews     : {completed_reviews}")
print(f"Selected Candidates   : {selected_candidates}")
print(f"Rejected Candidates   : {rejected_candidates}")
print(f"Shortlisted           : {shortlisted_candidates}")
print(f"In Review             : {in_review_candidates}")
print(f"Average Rating        : {average_rating}")
print(f"Average Experience    : {average_experience} Years")

# =====================================================
# TOP RECRUITERS
# =====================================================

print("\nTop Recruiters")

top_recruiters = (
    df.groupby("Recruiter_Name")
    .size()
    .reset_index(name="Candidates_Handled")
    .sort_values(by="Candidates_Handled", ascending=False)
)

display(top_recruiters)

# =====================================================
# COMPANY SUMMARY
# =====================================================

print("\nCompany Summary")

company_summary = (
    df.groupby("Company")
    .size()
    .reset_index(name="Total_Candidates")
    .sort_values(by="Total_Candidates", ascending=False)
)

display(company_summary)

# =====================================================
# SAVE CLEAN DATASET
# =====================================================

output_file = "Recruiter_Analysis_Report.xlsx"

df.to_excel(output_file, index=False)

print(f"\nReport Saved Successfully : {output_file}")

# =====================================================
# FINAL MESSAGE
# =====================================================

print("\nProject Status : COMPLETED")
print("Task 18 - Admin Console & Review Queue Completed Successfully")
print("Dashboard, Analysis and Report Generated Successfully.")
```
